# Deep Agents Overview

A **deep agent** is an AI agent designed for complex, multi-step work. Instead of answering every request in one model call, it can create a plan, use tools, save intermediate work, and delegate focused tasks to other agents.

The `deepagents` library is a standalone framework built on LangGraph. It provides a structured runtime for agents that need more than a simple prompt-and-response loop, similar to coding assistants, research agents, and task automation systems.

This notebook introduces the main building blocks:

- **Planning:** breaking a broad request into smaller, trackable steps.
- **Tools:** Python functions the agent can call to interact with external systems, such as web search.
- **File-system tools:** reading and writing working notes so large results do not stay entirely in the conversation context.
- **Subagents:** specialized agents that handle an isolated part of a larger task.
- **Model:** the language model that decides what to say and which tools to use.

The examples use LangChain for model and agent abstractions, Tavily for web search, and environment variables for configuration.

## When to Use Deep Agents

Use a deep agent when a task has enough complexity that the model benefits from an explicit workflow. Typical signals include:

- **Multi-step reasoning:** the request must be planned and decomposed rather than answered directly.
- **Large context:** research notes, files, or tool results may exceed the useful conversation context.
- **Delegation:** separate subtasks can be assigned to specialized subagents.
- **Persistent work:** intermediate files or memory should be available across steps or conversations.

For a short question or a single deterministic function call, a regular model or basic agent is usually simpler and faster.

The setup cell loads secrets from `config/.env`. Create that file locally with values for `GOOGLE_API_KEY`, `GROQ_API_KEY`, and `TAVILY_API_KEY`, plus the model names stored in `GOOGLE_MODEL` and `GROQ_MODEL`. Keep this file out of source control.

## What Happened?

When you invoked the deep agent, it returned a **state**: a dictionary containing the conversation and any working artifacts produced during the run. The exact state keys depend on the configured tools and runtime, but the examples above inspect two useful parts:

- `result["messages"]`: the ordered conversation history, including the user request, tool calls, tool results, and the final response.
- `result["files"]`: files created or updated by the agent while it manages its working context.

During the research task, the agent may automatically:

1. **Plan:** use the built-in `write_todos` tool to break the request into steps.
2. **Research:** call the `web_search` or `internet_search` tool to gather information.
3. **Manage context:** use file-system tools such as `write_file` and `read_file` to offload large results.
4. **Delegate:** spawn a subagent when an isolated specialist task is useful.
5. **Synthesize:** combine the intermediate findings into a final response.

Tool use is model-directed, so the exact sequence can vary between runs.

## System Prompt

A **system prompt** is the high-priority instruction that defines an agent's role, goals, boundaries, and working style. It is different from the user's request: the user asks for a specific outcome, while the system prompt explains how the agent should approach that outcome.

Deep agents include a built-in system prompt inspired by Claude Code. It describes how to use the planning tool, file-system tools, and subagents. Add a custom system prompt when the agent needs domain-specific behavior, such as producing research reports, following a review checklist, or using a particular output format.

A good custom prompt complements the built-in capabilities. It should state the role and quality bar clearly without duplicating every tool instruction.

```python
from deepagents import create_deep_agent

research_instructions = """\
You are an expert researcher. Your job is to conduct \
thorough research, and then write a polished report. \
"""

agent = create_deep_agent(
    model=model,
    system_prompt=research_instructions,
)
result = agent.invoke({"messages": [{"role": "user", "content": "What is deepagent?"}]})
result
```
